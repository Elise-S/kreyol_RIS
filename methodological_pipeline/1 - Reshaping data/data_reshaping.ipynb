### 2. Removal of unnecessary info, lang ID column added, concatenation of dataframe per corpus
#### a. Removal of all columns except text/content + iso code column added

```python
# fineweb2
print(list(df_fnw_gcf.columns.values))
```

```python
cols_to_remove = ["id", "media", "metadata"]
df_fnw_gcf = df_fnw_gcf.drop(columns=cols_to_remove)
df_fnw_gcr = df_fnw_gcr.drop(columns=cols_to_remove)
df_fnw_hat = df_fnw_hat.drop(columns=cols_to_remove)
```

```python
df_fnw_gcf["iso-639-3"] = "gcf"
df_fnw_gcr["iso-639-3"] = "gcr"
df_fnw_hat["iso-639-3"] = "hat"
```

```python
df_fnw_gcr
```

```python
# Glotcc
cols = ['warc-target-uri', 'warc-date', 'warc-record-id', 'quality-warnings', 'categories', 'identification-language', 'identification-prob', 'identification-consistency', 'script-percentage', 'num-sents', 'content-length', 'tlsh']
df_glotcc_gcf = df_glotcc_gcf.drop(columns=cols)
df_glotcc_gcr = df_glotcc_gcr.drop(columns=cols)
df_glotcc_hat = df_glotcc_hat.drop(columns=cols)
```

```python
df_glotcc_gcf
```

```python
df_glotcc_gcf["iso-639-3"] = "gcf"
df_glotcc_gcr["iso-639-3"] = "gcr"
df_glotcc_hat["iso-639-3"] = "hat"
```

```python
df_glotcc_hat
```

#### b. Concatenation per corpus + adding of corpus column

```python
# concatenation fnw
df_fnw = pd.concat([df_fnw_gcf, df_fnw_gcr, df_fnw_hat], ignore_index=True)
df_fnw
```

```python
df_fnw["original-corpus"] = "fineweb2"
df_fnw
```

```python
# concatenation glotcc
df_glotcc = pd.concat([df_glotcc_gcf, df_glotcc_gcr, df_glotcc_hat], ignore_index=True)
df_glotcc["original-corpus"] = "glotcc"
df_glotcc
```

```python
# final concatenation of the two dataframes 
df_glotcc = df_glotcc.rename(columns={"content": "text"}) # change glotcc 'content' column header into 'text', to align with fineweb2
df_fnw_glotcc = pd.concat([df_fnw, df_glotcc], ignore_index=True)
df_fnw_glotcc
```

```python
print(df_fnw_glotcc["iso-639-3"].value_counts())
```

```python
print(df_fnw_glotcc["original-corpus"].value_counts())
```

```python
# adding index as columns
df_fnw_glotcc.reset_index(level=0, inplace=True)
df_fnw_glotcc
```

```python
df_fnw_glotcc.to_csv("fineweb2_glotcc_4C.csv", index=False)
```

### 3. Spacy - Count the number of sentences per language

```python
# count split lang and count text per lang before
nlp = spacy.blank("xx")
nlp.add_pipe("sentencizer")
```

```python
# Group texts by iso-code and concatenate into a single string
grouped = df_fnw_glotcc.groupby("iso-639-3")["text"].apply(" ".join)
print(grouped)
```

```python
doc_gcf = nlp(grouped["gcf"])
doc_gcr = nlp(grouped["gcr"])
doc_hat = nlp(grouped["hat"])
print(type(doc_hat))
```

```python
print(f"Number of sentences for gcf: {len(list(doc_gcf.sents))} \nNumber of sentences for gcr: {len(list(doc_gcr.sents))} \nNumber of sentences for hat: {len(list(doc_hat.sents))}")
```

```python
# another try with blank fr.
nlp = spacy.blank("fr")
nlp.add_pipe("sentencizer")
doc_gcf = nlp(grouped["gcf"])
doc_gcr = nlp(grouped["gcr"])
doc_hat = nlp(grouped["hat"])
print(f"Number of sentences for gcf: {len(list(doc_gcf.sents))} \nNumber of sentences for gcr: {len(list(doc_gcr.sents))} \nNumber of sentences for hat: {len(list(doc_hat.sents))}")
```

```python
# last try with little French model
nlp = spacy.load("fr_core_news_sm")
doc_gcf = nlp(grouped["gcf"])
doc_gcr = nlp(grouped["gcr"])
doc_hat = nlp(grouped["hat"])
print(f"Number of sentences for gcf: {len(list(doc_gcf.sents))} \nNumber of sentences for gcr: {len(list(doc_gcr.sents))} \nNumber of sentences for hat: {len(list(doc_hat.sents))}")
```
