```python
# evaluation of the original tags -> collapsed 3-class precision/recall/F1 (0 granularity for gcf cases -> always said correct)
from sklearn.metrics import classification_report, confusion_matrix

collapse = {"gcf": "gcf", "gcf-mar": "gcf", "gcf-gua": "gcf", "gcr": "gcr", "hat": "hat"}
gold = [collapse[l] for l in glottocode_tag_list]
basic = original_code_list

print(classification_report(gold, basic, labels=["gcf", "gcr", "hat"]))
print(confusion_matrix(gold, basic, labels=["gcf", "gcr", "hat"]))
```

```python
# Evaluation of the original tags -> hierarchical Precision/Recall/F1

from collections import Counter

# Defining parent relationships (a label's ancestors, not including itself) -> gold tags as ancestors, ancestors as keys in the dict
parents = {
    "gcf": [],
    "gcf-mar": ["gcf"],
    "gcf-gua": ["gcf"],
    "gcr": [],
    "hat": [],
}

def expand(label):
    """Return the set of a label plus all its ancestors."""
    result = {label}
    to_process = [label]
    while to_process:
        current = to_process.pop()
        for p in parents.get(current, []):
            if p not in result:
                result.add(p)
                to_process.append(p)
    return result

def hierarchical_prf(basic_labels, gold_labels):
    """
    basic_labels, gold_labels: parallel lists of string labels, one per document.
    Returns (precision, recall, f1) computed over the whole set (micro-averaged,
    the standard way hierarchical P/R/F1 is reported).
    """
    numerator = 0
    basic_total = 0
    gold_total = 0

    for b, g in zip(basic_labels, gold_labels):
        exp_b = expand(b)
        exp_g = expand(g)
        numerator += len(exp_b & exp_g)
        basic_total += len(exp_b)
        gold_total += len(exp_g)

    hP = numerator / basic_total if basic_total else 0.0
    hR = numerator / gold_total if gold_total else 0.0
    hF1 = 2 * hP * hR / (hP + hR) if (hP + hR) else 0.0

    return hP, hR, hF1

hP, hR, hF1 = hierarchical_prf(basic, gold)
print(f"Hierarchical Precision: {hP:.3f}")
print(f"Hierarchical Recall:    {hR:.3f}")
print(f"Hierarchical F1:        {hF1:.3f}")
```

```python
# Evaluation of the original tags -> hierarchical evaluation per category (all)

def hierarchical_prf_by_gold_class(basic_labels, gold_labels):
    by_class = {}
    classes = sorted(set(gold_labels))
    for cls in classes:
        idxs = [i for i, g in enumerate(gold_labels) if g == cls]
        b_sub = [basic_labels[i] for i in idxs]
        g_sub = [gold_labels[i] for i in idxs]
        by_class[cls] = hierarchical_prf(b_sub, g_sub)
    return by_class

for cls, (p, r, f1) in hierarchical_prf_by_gold_class(basic, glottocode_tag_list).items():
    print(f"{cls:10s}  hP={p:.3f}  hR={r:.3f}  hF1={f1:.3f}")
```

```python
# Evaluation of the original tags -> hierarchical evaluation per category (gcf-mar & gcf-gua only)

def hierarchical_prf_filtered(basic_labels, gold_labels, gold_subset):
    idxs = [i for i, g in enumerate(gold_labels) if g in gold_subset]
    b_sub = [basic_labels[i] for i in idxs]
    g_sub = [gold_labels[i] for i in idxs]
    return hierarchical_prf(b_sub, g_sub)

hP, hR, hF1 = hierarchical_prf_filtered(basic, glottocode_tag_list, {"gcf-mar", "gcf-gua"})
print(f"gcf subtypes combined — hP={hP:.3f}  hR={hR:.3f}  hF1={hF1:.3f}")
```

```python
# fixing unicode and removing some punctuation strings
import ftfy

punct_to_remove = [" [...] ", "\n[...]\n", "[...]\n", " [...]", "[...] ",  "\n[...]", "[...]"]

def remove_punct (punct_to_remove, corpus):
    for punct in punct_to_remove:
        clean_text = corpus.replace(punct, "")
        text = ftfy.fix_text(clean_text)
    return text

text_gdl = remove_punct(punct_to_remove, corpus_gdl)
text_mqc = remove_punct(punct_to_remove, corpus_mqc)
text_gfr = remove_punct(punct_to_remove, corpus_gfr)
text_hcr = remove_punct(punct_to_remove, corpus_hcr)

print(text_gdl)
```
