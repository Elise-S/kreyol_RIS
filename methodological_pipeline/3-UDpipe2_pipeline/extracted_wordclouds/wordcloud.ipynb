```python
import numpy as np
import pandas as pd
from os import path
from PIL import Image
from wordcloud import WordCloud, STOPWORDS, ImageColorGenerator
```

```python
import matplotlib.pyplot as plt
```

```python
# Import files
def open_file(file_path):
    with open(file_path, "r", encoding="utf-8", errors='ignore') as f:
        content = f.read()
    return content

text_gdl = open_file(r"gdl_guadeloupe_corpus.txt")
text_gfr = open_file(r"gfr_guyane_corpus.txt")
text_hcr = open_file(r"hcr_haiti_corpus.txt")
text_mqc = open_file(r"mqc_martinique_corpus.txt")
```

```python
print(text_gdl)
```

```python
# Word cloud image generation for gdl:
wordcloud_gdl = WordCloud(max_font_size=50, max_words=60, background_color="white").generate(text_gdl)

# Display the generated image:
plt.figure()
plt.imshow(wordcloud_gdl, interpolation='bilinear')
plt.axis("off")
plt.show()
```

```python
# Word cloud image generation for mqc:
wordcloud_mqc = WordCloud(max_font_size=50, max_words=60, background_color="white").generate(text_mqc)

# Display the generated image:
plt.figure()
plt.imshow(wordcloud_mqc, interpolation='bilinear')
plt.axis("off")
plt.show()
```

```python
# Word cloud image generation for gdl:
wordcloud_hcr = WordCloud(max_font_size=50, max_words=60, background_color="white").generate(text_hcr)

# Display the generated image:
plt.figure()
plt.imshow(wordcloud_hcr, interpolation='bilinear')
plt.axis("off")
plt.show()
```

```python
# Word cloud image generation for gdl:
wordcloud_gfr = WordCloud(max_font_size=50, max_words=60, background_color="white").generate(text_gfr)

# Display the generated image:
plt.figure()
plt.imshow(wordcloud_gfr, interpolation='bilinear')
plt.axis("off")
plt.show()
```

```python
# Save images:
wordcloud_gdl.to_file("wordcloud_gdl.png")
wordcloud_mqc.to_file("wordcloud_mqc.png")
wordcloud_hcr.to_file("wordcloud_hcr.png")
wordcloud_gfr.to_file("wordcloud_gfr.png")
```
