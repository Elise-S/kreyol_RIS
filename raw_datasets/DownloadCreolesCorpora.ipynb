````python
# Libraries

!pip install datatrove
from datatrove.pipeline.readers import ParquetReader
import pandas as pd
````

````python
# Importation from HuggingFace of the train and tests splits of the subset gcf_Latn of the corpus Fineweb-2
# The "limit" parameter determines how many documents will be streamed, fixed here to 25
# This will fetch the Guadeloupean/Martinican Creoles filtered data

# Importation with ParquetReader
data_reader_gcf_train = ParquetReader("hf://datasets/HuggingFaceFW/fineweb-2/data/gcf_Latn/train", limit=25)
data_reader_gcf_test = ParquetReader("hf://datasets/HuggingFaceFW/fineweb-2/data/gcf_Latn/test", limit=25)

# Reformating of the two data splits into DataFrames
df_gcf_train = pd.DataFrame([doc for doc in data_reader_gcf_train()])
df_gcf_test = pd.DataFrame([doc for doc in data_reader_gcf_test()])

# Observation of the two DataFrames
print(df_gcf_train.shape)
print(df_gcf_train.head())
print(df_gcf_test.shape)
print(df_gcf_test.head())

# Exportation into CSV files
df_gcf_train.to_csv("fineweb2_gcf_train.csv", index=False)
df_gcf_test.to_csv("fineweb2_gcf_test.csv", index=False)
````

````python
# Combine the train and test splits into a same DataFrame
df_gcf = pd.concat([df_gcf_train, df_gcf_test], ignore_index=True)

# Check
print(df_gcf.shape)
print(df_gcf.head())

# Exportation into CSV file
df_gcf.to_csv("fineweb2_gcf.csv", index=False)
````

````python
# Libraries

from datasets import load_dataset
import pandas as pd
````

````python
# Importation from HuggingFace of the clean split of the subset gcf of the corpus MADLAD-400
# Method: importation with dataset.load_dataset
# This will fetch the Guadeloupean/Martinican Creoles resulting data
# Dataset limited to 25 documents

# Importation
dataset_madlad_gcf_clean = load_dataset("allenai/madlad-400", "gcf", split="clean")

# Reformating into dataframe
df_dataset_madlad_gcf = pd.DataFrame(dataset_madlad_gcf_clean)

# Only keeping the 25 first rows
df_madlad_gcf = df_dataset_madlad_gcf[df_dataset_madlad_gcf.index <= 24]

# Check
print(df_madlad_gcf.shape)
print(df_madlad_gcf.head())

# Exportation into CSV files
df_madlad_gcf.to_csv("madlad_gcf.csv", index=False)
````
